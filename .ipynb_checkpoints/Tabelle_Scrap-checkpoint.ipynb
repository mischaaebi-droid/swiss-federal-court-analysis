```python
import pandas as pd
import requests
import certifi
import re
pd.set_option("display.max_rows", 200)

from io import StringIO

url = "https://de.wikipedia.org/wiki/Liste_der_Mitglieder_des_Schweizer_Bundesgerichts"

response = requests.get(
    url,
    headers={"User-Agent": "Mozilla/5.0"},
    verify=certifi.where(),
    timeout=30
)

response.raise_for_status()

tabellen = pd.read_html(StringIO(response.text))

print(f"{len(tabellen)} Tabellen gefunden")

df = tabellen[0]



```

```python
df["Nachname"] = df["Name"].str.extract(r"\w+\s+(.*)")
df
```

```python
df["Wahl"] = pd.to_numeric(df["Wahl"], errors="coerce")

df[df["Wahl"]> 2006].to_excel("bundesrichter_ab_2007.xlsx", index=False)
```
