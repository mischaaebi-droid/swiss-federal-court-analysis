```python
import pandas as pd
pd.set_option("display.max_colwidth", None)
pd.set_option("display.max_rows", 200)
df=pd.read_csv("decisions_data_structured.csv")
df.head(2)
```

## Cosideration of decisions made by 5 judges.

```python
richter_pro_urteil = df.groupby("decision_sign").size()

fuenfer_urteile = richter_pro_urteil[
    richter_pro_urteil == 5
].index

df_5 = df[df["decision_sign"].isin(fuenfer_urteile)]

total_number_decicion_with_5_judges = len(fuenfer_urteile)

print("Total number of decisions with 5 Judges: "+ str(total_number_decicion_with_5_judges))
```

 ### 1.Comparision between decisions of left- and right-wing-judge-teams
* right-wing-judgeteam = at least 2 right-wing-judges (SP or Grüne)
* right-wing-judgeteam = at least 2 right-wing-judges (SVP or FDP)

#### **1a) Comparisen over all legal areas**

```python
# 1. Pro Urteil zählen, wie viele Richter SVP oder FDP sind
right_major = (df_5["party"].isin(["SVP", "FDP"]).groupby(df_5["decision_sign"]).sum()>= 3)
left_major = (df_5["party"].isin(["SP", "Grüne"]).groupby(df_5["decision_sign"]).sum()>= 3)
center_major = (df_5["party"].isin(["Die Mitte", "GLP"]).groupby(df_5["decision_sign"]).sum() >= 3)

center_right_major = (df_5["party"].isin(["SVP", "FDP", "Die Mitte", "GLP"]).groupby(df_5["decision_sign"]).sum()>= 3)
center_left_major = (df_5["party"].isin(["SP", "Grüne", "Die Mitte", "GLP"]).groupby(df_5["decision_sign"]).sum()>= 3)


without_right = (df_5["party"].isin(["FDP", "SVP"]).groupby(df_5["decision_sign"]).sum() == 0)
without_left = (df_5["party"].isin(["SP", "Grüne"]).groupby(df_5["decision_sign"]).sum() == 0)

# 2. Nur die Urteile mit rechter Mehrheit herausnehmen
right_wing_decisions = right_major[right_major].index
left_wing_decisions = left_major[left_major].index
center_decisions = center_major[center_major].index

center_right_decisions = center_right_major.index
center_left_decisions = center_left_major.index

without_right_decisions = without_left[without_left].index
without_left_decisions = without_left[without_left].index

# 5. Von diesen Urteilen zählen, wie viele gutgeheissen
# oder teilweise gutgeheissen wurden
number_appeal_success_right = df_5[df_5["decision_sign"].isin(right_wing_decisions)& df_5["entscheid"].isin(["gutgeheissen","teilweise gutgeheissen"])]["decision_sign"].nunique()
number_appeal_success_left = df_5[df_5["decision_sign"].isin(left_wing_decisions)& df_5["entscheid"].isin(["gutgeheissen","teilweise gutgeheissen"])]["decision_sign"].nunique()
number_appeal_success_center = df_5[df_5["decision_sign"].isin(center_decisions)& df_5["entscheid"].isin(["gutgeheissen", "teilweise gutgeheissen"])]["decision_sign"].nunique()

# Calculate all variables
total_number_decicion_with_5_judges = len(df_5) / 5

anzahl_right_wing_decisions = right_major.sum()
anzahl_left_wing_decisions = left_major.sum()
anzahl_center_decisions = center_major.sum()
anzahl_center_right_decisions = center_right_major.sum()
anzahl_center_left_decisions = center_left_major.sum()
anzahl_without_right_decisions = without_right.sum()
anzahl_without_left_decisions = without_left.sum()

print("NUMBERS OF DECISIONS")
print("total_5-judge-panel-decisions: " + str(total_number_decicion_with_5_judges))

print("How 5-judge-decisions are made-------------------------")
print("Right-wing decisions:", anzahl_right_wing_decisions)
print("Left-wing decisions:", anzahl_left_wing_decisions)
print("Center decisions:", anzahl_center_decisions)
print("Center-right decisions:", anzahl_center_right_decisions)
print("Center-left decisions:", anzahl_center_left_decisions)
print("Decisions without right:", anzahl_without_right_decisions)
print("Decisions without left:", anzahl_without_left_decisions)

```
