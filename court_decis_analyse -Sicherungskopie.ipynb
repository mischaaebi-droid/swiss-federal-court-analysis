```python
import pandas as pd
pd.set_option("display.max_colwidth", None)
pd.set_option("display.max_rows", 200)
df=pd.read_csv("decisions_data_structured.csv")
df.head(2)
```

## Cosideration of decisions made by 3 judges.

```python
richter_pro_urteil = df.groupby("decision_sign").size()

dreier_urteile = richter_pro_urteil[
    richter_pro_urteil == 3
].index

df_3 = df[df["decision_sign"].isin(dreier_urteile)]

total_number_decicion_with_3_judges = len(dreier_urteile)

print("Total number of decisions with 3 Judges: "+ str(total_number_decicion_with_3_judges))
```

 ### 1.Comparision between decisions of left- and right-wing-judge-teams
* right-wing-judgeteam = at least 2 right-wing-judges (SP or Grüne)
* right-wing-judgeteam = at least 2 right-wing-judges (SVP or FDP)

#### **1a) Comparisen over all legal areas**

```python
# 1. Pro Urteil zählen, wie viele Richter SVP oder FDP sind
right_major = (df_3["party"].isin(["SVP", "FDP"]).groupby(df_3["decision_sign"]).sum()>= 2)
left_major = (df_3["party"].isin(["SP", "Grüne"]).groupby(df_3["decision_sign"]).sum()>= 2)

# 2. Nur die Urteile mit rechter Mehrheit herausnehmen
right_wing_decisions = right_major[right_major].index
left_wing_decisions = left_major[left_major].index

# 3. Von diesen Urteilen zählen, wie viele gutgeheissen
# oder teilweise gutgeheissen wurden
number_appeal_success_right = df_3[df_3["decision_sign"].isin(right_wing_decisions)& df_3["entscheid"].isin(["gutgeheissen","teilweise gutgeheissen"])]["decision_sign"].nunique()
number_appeal_success_left = df_3[df_3["decision_sign"].isin(left_wing_decisions)& df_3["entscheid"].isin(["gutgeheissen","teilweise gutgeheissen"])]["decision_sign"].nunique()

# Calculate all variables
total_number_decicion_with_3_judges = len(df_3) / 3
anzahl_right_wing_decisions = right_major.sum()
anzahl_left_wing_decisions = left_major.sum()

percentage_right_wing_decisions = (anzahl_right_wing_decisions / total_number_decicion_with_3_judges * 100)
percentage_left_wing_decisions = (anzahl_left_wing_decisions / total_number_decicion_with_3_judges * 100)


prercent_right_wing = ( number_appeal_success_right / anzahl_right_wing_decisions * 100)
prercent_left_wing = (number_appeal_success_left / anzahl_left_wing_decisions * 100)

print("NUMBERS OF DECISIONS")
print("total_3-judge-panel-decisions: " + str(total_number_decicion_with_3_judges))

print("Number of rightwing_decision: " + str(anzahl_right_wing_decisions))
print("Number of leftwing_decision: " + str(anzahl_left_wing_decisions))
print(f"Percentage of decisions with right-wing majority: {percentage_right_wing_decisions:.1f}%")
print(f"Percentage of decisions with left-wing majority: {percentage_left_wing_decisions:.1f}%")
print("---------------------------------------------")
print("SUCCESS")
print("Number of success with right majority: " + str(number_appeal_success_right))
print("Number of success with left majority: " + str(number_appeal_success_left))

print(f"Success-rate in cases with a right-wing-majority: {prercent_right_wing:.1f}%")
print(f"Success-rate in cases with a left-wing-majority: {prercent_left_wing:.1f}%")
```

#### **1b) Comparisen only Strafrecht (=Criminal Cases)**

```python
df_3_strafrecht=df_3[df_3["legal area"]=="Strafrecht"]
```

```python
# 1. Pro Urteil zählen, wie viele Richter SVP oder FDP sind
right_major_crim = (df_3_strafrecht["party"].isin(["SVP", "FDP"]).groupby(df_3_strafrecht["decision_sign"]).sum()>= 2)
left_major_crim = (df_3_strafrecht["party"].isin(["SP", "Grüne"]).groupby(df_3_strafrecht["decision_sign"]).sum()>= 2)


number_appeal_success_right_crim = df_3_strafrecht[df_3_strafrecht["decision_sign"].isin(right_wing_decisions)& df_3_strafrecht["entscheid"].isin(["gutgeheissen","teilweise gutgeheissen"])]["decision_sign"].nunique()
number_appeal_success_left_crim = df_3_strafrecht[df_3_strafrecht["decision_sign"].isin(left_wing_decisions)& df_3_strafrecht["entscheid"].isin(["gutgeheissen","teilweise gutgeheissen"])]["decision_sign"].nunique()

anzahl_right_wing_decisions_crim = right_major_crim.sum()
anzahl_left_wing_decisions_crim = left_major_crim.sum()

prercent_right_wing_crim = (number_appeal_success_right_crim / anzahl_right_wing_decisions_crim* 100)
prercent_left_wing_crim = (number_appeal_success_left_crim / anzahl_left_wing_decisions_crim * 100)

print("ONLY CRIMINAL CASES, ONLY 3-JUDGE TEAMS:")
print("Number of rightwing_decisions: " +str(anzahl_right_wing_decisions_crim))
print("Number of leftwing_decisions: " +str(anzahl_left_wing_decisions_crim))

print("Number of success with right majority: " + str(number_appeal_success_right_crim))
print("Number of success with left majority: " + str(number_appeal_success_left_crim))
print(f"Success-rate with a right-wing-majority: {prercent_right_wing_crim:.1f}%")
print(f"Success-rate with a left-wing-majority: {prercent_left_wing_crim:.1f}%")
```
